## 1. Variables and types

Java is **statically typed**: every variable has a type that can't change.

| Type | Example | Use |
|---|---|---|
| `int` | `42` | whole numbers |
| `double` | `3.14` | decimal numbers |
| `boolean` | `true` | yes/no |
| `char` | `'A'` | one character |
| `String` | `"Hello"` | text (a class, not a primitive) |

```java
int age = 20;
double height = 1.75;
boolean isStudent = true;
char initial = 'A';
String name = "Ada";

System.out.println(name + " is " + age + " years old");
System.out.printf("Height: %.2f m, initial: %c, student: %b%n", height, initial, isStudent);
```

```java
var city = "London";     // var lets Java infer the type (still a String)
final double PI = 3.14159; // final: a constant, can't be changed
System.out.println(city.length() + " letters, PI = " + PI);
```

## 2. Operators

Watch out: dividing two `int`s gives an `int` (the decimals are cut off). Cast to `double` when you need decimals.

```java
System.out.println(7 / 2);           // 3
System.out.println(7 % 2);           // 1  (remainder)
System.out.println((double) 7 / 2);  // 3.5
System.out.println(age >= 18 && isStudent);
int counter = 0;
counter++;
counter += 5;
System.out.println(counter);
```

## 3. Control flow

`if`/`else`, `switch`, and the loops `for`, `while` and `do … while`. Blocks use `{ }`.

```java
double grade = 7.5;

if (grade >= 9) {
    System.out.println("Excellent");
} else if (grade >= 5) {
    System.out.println("Pass");
} else {
    System.out.println("Fail");
}

String day = "SAT";
String type = switch (day) {
    case "SAT", "SUN" -> "weekend";
    default -> "weekday";
};
System.out.println(day + " is a " + type);
```

```java
for (int i = 1; i <= 5; i++) {
    System.out.println(i + " squared is " + (i * i));
}

int n = 10, steps = 0;
while (n != 1) {
    n = (n % 2 == 0) ? n / 2 : 3 * n + 1;
    steps++;
}
System.out.println("Reached 1 in " + steps + " steps");
```

## 4. Methods

A method has a return type (or `void`), a name and typed parameters. Methods with the same name but different parameters are **overloaded**.

```java
int area(int width, int height) {
    return width * height;
}

double area(double radius) {          // overload: same name, different parameters
    return Math.PI * radius * radius;
}

System.out.println(area(3, 4));
System.out.println(area(1.0));
```

## 5. Arrays and ArrayList

- An **array** has a fixed size: `int[] marks = new int[5];`
- An **`ArrayList`** grows and shrinks as needed and only holds objects (`Integer`, not `int`).

```java
int[] marks = {6, 8, 9, 4, 7};
int sum = 0;
for (int mark : marks) {
    sum += mark;
}
System.out.println("Count: " + marks.length + ", average: " + (double) sum / marks.length);
System.out.println(java.util.Arrays.toString(marks));
```

```java
import java.util.ArrayList;
import java.util.List;

List<String> names = new ArrayList<>();
names.add("Ada");
names.add("Alan");
names.add("Grace");
names.remove("Alan");
System.out.println(names + " size=" + names.size() + " first=" + names.get(0));
```

## 6. Classes and objects

A **class** is a blueprint; **objects** are built from it with `new`.
**Encapsulation**: keep fields `private` and give controlled access through methods (getters/setters).

```java
class BankAccount {
    private final String owner;
    private double balance;

    BankAccount(String owner, double initialBalance) {   // constructor
        this.owner = owner;
        this.balance = initialBalance;
    }

    void deposit(double amount) {
        if (amount <= 0) {
            throw new IllegalArgumentException("Deposit must be positive");
        }
        balance += amount;
    }

    double getBalance() { return balance; }
    String getOwner() { return owner; }

    @Override
    public String toString() {
        return owner + ": " + balance + " EUR";
    }
}

BankAccount account = new BankAccount("Ada", 100);
account.deposit(50);
System.out.println(account);
// account.balance = 1000000;   // not allowed: balance is private
```

## 7. Inheritance and interfaces

- A class can **extend** one other class and reuse its code. `super(...)` calls the parent's constructor.
- An **interface** is a contract: a list of methods a class promises to have. A class can **implement** many interfaces.
- `abstract` classes can't be created directly; they leave some methods for subclasses to define.

```java
interface Shape {
    double area();
    default String describe() {
        return getClass().getSimpleName() + " with area " + String.format("%.2f", area());
    }
}

abstract class NamedShape implements Shape {
    protected final String label;
    NamedShape(String label) { this.label = label; }
}

class Circle extends NamedShape {
    private final double radius;
    Circle(double radius) { super("circle"); this.radius = radius; }
    public double area() { return Math.PI * radius * radius; }
}

class Rectangle extends NamedShape {
    protected final double width, height;
    Rectangle(double width, double height) { super("rectangle"); this.width = width; this.height = height; }
    public double area() { return width * height; }
}

class Square extends Rectangle {
    Square(double side) { super(side, side); }
    @Override
    public String describe() { return "Square! " + super.describe(); }
}
```

## 8. Polymorphism

A variable of type `Shape` can hold *any* object that implements `Shape`. When you call `area()` or `describe()`,
Java runs the version of the **actual object**: that's polymorphism.

```java
List<Shape> shapes = List.of(new Circle(1), new Rectangle(2, 3), new Square(2));

double total = 0;
for (Shape shape : shapes) {
    System.out.println(shape.describe());
    total += shape.area();
}
System.out.printf("Total area: %.2f%n", total);

Shape s = shapes.get(2);
if (s instanceof Square sq) {          // check the real type and cast in one step
    System.out.println("It's a square of width " + sq.width);
}
```

## 9. Exceptions

Errors are **exceptions**. Use `try` / `catch` to handle them and `finally` for code that must always run.
You can create your own exception classes by extending `Exception` (checked: must be declared with `throws`)
or `RuntimeException` (unchecked).

```java
try {
    int[] small = new int[2];
    small[5] = 1;
} catch (ArrayIndexOutOfBoundsException e) {
    System.out.println("Oops: " + e.getMessage());
} finally {
    System.out.println("finally always runs");
}

try {
    account.deposit(-10);
} catch (IllegalArgumentException e) {
    System.out.println("Rejected: " + e.getMessage());
}
```

```java
class InsufficientFundsException extends Exception {
    InsufficientFundsException(String message) { super(message); }
}

void checkWithdrawal(BankAccount acc, double amount) throws InsufficientFundsException {
    if (amount > acc.getBalance()) {
        throw new InsufficientFundsException(acc.getOwner() + " only has " + acc.getBalance());
    }
    System.out.println("OK, " + acc.getOwner() + " can withdraw " + amount);
}

try {
    checkWithdrawal(account, 20);
    checkWithdrawal(account, 1000);
    System.out.println("never printed: the line above threw an exception");
} catch (InsufficientFundsException e) {
    System.out.println("Can't withdraw: " + e.getMessage());
}
```
