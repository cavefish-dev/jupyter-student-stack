## 1. Variables and types

Python works out the type of a value by itself. The basic types are `int`, `float`, `str` (text) and `bool` (`True`/`False`).
Use `type()` to check a value's type.

````python
age = 20
height = 1.75
name = "Ada"
is_student = True

print(name, "is", age, "years old")
print(type(age), type(height), type(name), type(is_student))
````

````python
# Operators
print(7 + 2, 7 - 2, 7 * 2, 7 / 2)   # / always gives a float
print(7 // 2, 7 % 2, 2 ** 10)       # integer division, remainder, power
print("Hello, " + name + "!")
print(f"{name} will be {age + 1} next year")   # f-strings insert values
````

## 3. Conditionals

Python uses **indentation** (4 spaces) instead of braces to group code.

````python
grade = 7.5

if grade >= 9:
    print("Excellent")
elif grade >= 5:
    print("Pass")
else:
    print("Fail")

print("adult" if age >= 18 else "minor")
````

## 4. Loops

- `for` goes through each item of a sequence. `range(n)` gives `0, 1, …, n-1`.
- `while` repeats while a condition is true.
- `break` exits a loop, `continue` jumps to the next iteration.

````python
for i in range(1, 6):
    print(i, "squared is", i ** 2)
````

````python
n = 10
steps = 0
while n != 1:
    n = n // 2 if n % 2 == 0 else 3 * n + 1
    steps += 1
print("Reached 1 in", steps, "steps")
````

## 5. Functions

Define functions with `def`. Parameters can have default values, and `return` gives back a result.

````python
def area(width, height=1):
    """Return the area of a rectangle."""
    return width * height

print(area(3, 4))
print(area(5))
print(area(height=2, width=6))
````

## 6. Lists and dictionaries

- A **list** is an ordered collection: `[1, 2, 3]`. Lists can change (add, remove, sort).
- A **dictionary** maps keys to values: `{"name": "Ada", "age": 20}`.

````python
marks = [6.5, 8, 9.5, 4]
marks.append(7)
print(marks, "length:", len(marks))
print("first:", marks[0], "last:", marks[-1], "first two:", marks[:2])
print("average:", sum(marks) / len(marks))

passed = [m for m in marks if m >= 5]   # list comprehension
print("passed:", passed)
````

````python
student = {"name": "Ada", "age": 20, "courses": ["Maths", "CS"]}
student["age"] += 1
student["city"] = "London"

for key, value in student.items():
    print(f"{key}: {value}")
````

## 7. Classes and objects

A **class** describes a kind of object: its data (attributes) and what it can do (methods).
`__init__` runs when an object is created; `self` is the object itself.

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "..."

    def describe(self):
        return f"{self.name} says {self.speak()}"


class Dog(Animal):          # Dog inherits from Animal
    def speak(self):
        return "Woof!"


class Cat(Animal):
    def speak(self):
        return "Meow!"


for pet in [Dog("Rex"), Cat("Tom"), Animal("Generic")]:
    print(pet.describe())   # each object uses its own speak(): polymorphism
````

## 8. Exceptions

When something goes wrong Python raises an **exception**. Catch it with `try` / `except` so your program can react instead of crashing.
`raise` throws your own exceptions; `finally` always runs.

````python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Can't divide by zero!")
        return None
    finally:
        print(f"(tried {a} / {b})")

print(safe_divide(10, 2))
print(safe_divide(1, 0))
````

````python
def set_age(value):
    if value < 0:
        raise ValueError("age can't be negative")
    return value

try:
    set_age(-5)
except ValueError as error:
    print("Error:", error)
````
