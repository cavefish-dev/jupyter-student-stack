## 2. Types

JavaScript values have types even if you don't write them: `number`, `string`, `boolean`, `undefined`, `null`, `object`…
Use `typeof` to check. Be careful with `==`: it converts types. Always use `===`.

```typescript
console.log(typeof 42, typeof 3.14, typeof "hi", typeof true, typeof undefined, typeof null);
console.log(7 / 2, 7 % 2, 2 ** 10);
console.log("5" == 5, "5" === 5);   // true false
console.log("5" + 1, "5" - 1);      // "51" 4  (surprise!)
```

## 3. Control flow

```typescript
const grade = 7.5;
if (grade >= 9) {
  console.log("Excellent");
} else if (grade >= 5) {
  console.log("Pass");
} else {
  console.log("Fail");
}
console.log(age >= 18 ? "adult" : "minor");
```

```typescript
for (let i = 1; i <= 5; i++) {
  console.log(`${i} squared is ${i * i}`);
}

let n = 10;
let steps = 0;
while (n !== 1) {
  n = n % 2 === 0 ? n / 2 : 3 * n + 1;
  steps++;
}
console.log("Reached 1 in", steps, "steps");
```

## 4. Functions and arrow functions

Functions are values: you can store them in variables and pass them to other functions.

```typescript
function area(width, height = 1) {
  return width * height;
}

const square = (x) => x * x;          // arrow function
const greet = (who) => {
  const message = `Hello, ${who}!`;
  return message;
};

console.log(area(3, 4), area(5), square(9), greet("world"));
```

## 5. Arrays and objects

Arrays have powerful methods such as `map`, `filter` and `reduce` that take a function.

```typescript
const marks = [6.5, 8, 9.5, 4];
marks.push(7);
console.log(marks, "length:", marks.length);

const passed = marks.filter((m) => m >= 5);
const doubled = marks.map((m) => m * 2);
const total = marks.reduce((acc, m) => acc + m, 0);
console.log({ passed, doubled, average: total / marks.length });
```

```typescript
const student = { name: "Ada", age: 20, courses: ["Maths", "CS"] };
student.city = "London";

for (const [key, value] of Object.entries(student)) {
  console.log(`${key}: ${value}`);
}

const { name: studentName, courses } = student;   // destructuring
console.log(studentName, courses[0]);
console.log(JSON.stringify(student));
```

## 6. Classes, inheritance and polymorphism

```typescript
class Animal {
  constructor(name) {
    this.name = name;
  }
  speak() {
    return "...";
  }
  describe() {
    return `${this.name} says ${this.speak()}`;
  }
}

class Dog extends Animal {
  speak() { return "Woof!"; }
}

class Cat extends Animal {
  speak() { return "Meow!"; }
}

for (const pet of [new Dog("Rex"), new Cat("Tom"), new Animal("Generic")]) {
  console.log(pet.describe());
}
```

## 7. Errors: `try` / `catch`

`throw` raises an error; `try` / `catch` handles it; `finally` always runs.

```typescript
function setAge(value) {
  if (value < 0) {
    throw new RangeError("age can't be negative");
  }
  return value;
}

try {
  setAge(-5);
} catch (error) {
  console.log(`${error.name}: ${error.message}`);
} finally {
  console.log("finally always runs");
}
```

## 8. A taste of TypeScript

TypeScript adds **type annotations**. The Deno kernel understands them directly. Types catch mistakes before the code runs
and make code easier to read.

```typescript
interface Shape {
  name: string;
  area(): number;
}

class Circle implements Shape {
  name = "circle";
  constructor(private radius: number) {}
  area(): number {
    return Math.PI * this.radius ** 2;
  }
}

class Rect implements Shape {
  name = "rectangle";
  constructor(private width: number, private height: number) {}
  area(): number {
    return this.width * this.height;
  }
}

function totalArea(shapes: Shape[]): number {
  return shapes.reduce((sum, s) => sum + s.area(), 0);
}

const shapes: Shape[] = [new Circle(1), new Rect(2, 3)];
shapes.forEach((s) => console.log(s.name, s.area().toFixed(2)));
console.log("total:", totalArea(shapes).toFixed(2));
```

## 9. Asynchronous code: `async` / `await`

Some operations take time (timers, network, files). `async` functions return a **Promise**, and `await` waits for it
without blocking everything else.

```typescript
const wait = (ms: number) => new Promise((resolve) => setTimeout(resolve, ms));

async function countdown(from: number) {
  for (let i = from; i > 0; i--) {
    console.log(i);
    await wait(200);
  }
  return "Lift off!";
}

await countdown(3);
```
