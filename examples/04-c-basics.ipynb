## 1. Program structure

- `#include <stdio.h>` gives you `printf` (and other input/output functions).
- Execution starts at `int main(void)`. Returning `0` means "everything went fine".
- Every statement ends with `;`.

````c
#include <stdio.h>

int main(void) {
    printf("Hello, world!\n");
    return 0;
}
````

## 2. Variables, types and `printf`

| Type | Example | `printf` format |
|---|---|---|
| `int` | `42` | `%d` |
| `double` | `3.14` | `%f` (`%.2f` for 2 decimals) |
| `char` | `'A'` | `%c` |
| text (`char` array) | `"Ada"` | `%s` |

`sizeof` tells you how many bytes a type uses.

````c
#include <stdio.h>

int main(void) {
    int age = 20;
    double height = 1.75;
    char initial = 'A';
    const char *name = "Ada";

    printf("%s (%c) is %d years old and %.2f m tall\n", name, initial, age, height);
    printf("7 / 2 = %d, 7 %% 2 = %d, 7.0 / 2 = %.1f\n", 7 / 2, 7 % 2, 7.0 / 2);
    printf("int: %zu bytes, double: %zu bytes, char: %zu byte\n", sizeof(int), sizeof(double), sizeof(char));
    return 0;
}
````

## 4. Control flow

````c
#include <stdio.h>

int main(void) {
    double grade = 7.5;
    if (grade >= 9) {
        printf("Excellent\n");
    } else if (grade >= 5) {
        printf("Pass\n");
    } else {
        printf("Fail\n");
    }

    for (int i = 1; i <= 5; i++) {
        printf("%d squared is %d\n", i, i * i);
    }

    int n = 10, steps = 0;
    while (n != 1) {
        n = (n % 2 == 0) ? n / 2 : 3 * n + 1;
        steps++;
    }
    printf("Reached 1 in %d steps\n", steps);

    int day = 6;
    switch (day) {
        case 6:
        case 7:
            printf("weekend\n");
            break;
        default:
            printf("weekday\n");
    }
    return 0;
}
````

## 5. Functions

A function must be declared before it's used. Arguments are passed **by value**: the function gets a copy.

````c
#include <stdio.h>

int area(int width, int height);   /* declaration (prototype) */

int main(void) {
    printf("area = %d\n", area(3, 4));
    return 0;
}

int area(int width, int height) {  /* definition */
    return width * height;
}
````

## 6. Arrays and strings

An array is a fixed-size block of values of the same type. A string is an array of `char` ending with `'\0'`.

````c
#include <stdio.h>
#include <string.h>

int main(void) {
    int marks[] = {6, 8, 9, 4, 7};
    int count = sizeof(marks) / sizeof(marks[0]);
    int sum = 0;
    for (int i = 0; i < count; i++) {
        sum += marks[i];
    }
    printf("%d marks, average %.2f\n", count, (double) sum / count);

    char word[20] = "Hello";
    strcat(word, ", C!");
    printf("%s has %zu characters\n", word, strlen(word));
    printf("first char: %c, code %d\n", word[0], word[0]);
    return 0;
}
````

## 7. Pointers

A **pointer** holds a memory **address**. `&x` is the address of `x`; `*p` is the value at the address in `p`.
Pointers let a function change the caller's variables, and they're how arrays really work.

````c
#include <stdio.h>

void swap(int *a, int *b) {
    int tmp = *a;
    *a = *b;
    *b = tmp;
}

int main(void) {
    int x = 1, y = 2;
    int *p = &x;
    printf("x = %d, address of x = %p, p points to %d\n", x, (void *) p, *p);

    *p = 10;               /* changes x through the pointer */
    printf("x is now %d\n", x);

    swap(&x, &y);
    printf("after swap: x = %d, y = %d\n", x, y);

    int numbers[] = {10, 20, 30};
    int *q = numbers;      /* an array name is a pointer to its first element */
    printf("%d %d %d\n", *q, *(q + 1), q[2]);
    return 0;
}
````

## 8. Structs

A `struct` groups related values into one type.

````c
#include <stdio.h>

struct Student {
    char name[30];
    int age;
    double grade;
};

void print_student(const struct Student *s) {
    printf("%s, %d years, grade %.1f\n", s->name, s->age, s->grade);   /* -> : field through a pointer */
}

int main(void) {
    struct Student class_list[] = {
        {"Ada", 20, 9.5},
        {"Alan", 22, 8.0},
    };
    class_list[1].grade += 0.5;
    for (int i = 0; i < 2; i++) {
        print_student(&class_list[i]);
    }
    return 0;
}
````

## 9. Dynamic memory: `malloc` and `free`

When you only know the size at run time, ask for memory with `malloc` (from `<stdlib.h>`).
**Always** check the result and `free` it when you're done, or your program leaks memory.

````c
#include <stdio.h>
#include <stdlib.h>

int main(void) {
    int n = 5;
    int *squares = malloc(n * sizeof(int));
    if (squares == NULL) {
        printf("Out of memory\n");
        return 1;
    }
    for (int i = 0; i < n; i++) {
        squares[i] = i * i;
    }
    for (int i = 0; i < n; i++) {
        printf("%d ", squares[i]);
    }
    printf("\n");
    free(squares);
    squares = NULL;
    return 0;
}
````
